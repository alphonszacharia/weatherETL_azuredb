```python
from pyspark.sql.functions import *

# Read Silver Delta table

silver_df = spark.table("silver.weather_cleaned")

display(silver_df)
```

```python
gold_df = silver_df \
    .withColumn(
        "weather_date",
        to_date("weather_time")
    ) \
    .groupBy(
        "location",
        "weather_date"
    ) \
    .agg(

        avg("temperature").alias("avg_temperature"),

        max("temperature").alias("max_temperature"),

        min("temperature").alias("min_temperature"),

        avg("humidity").alias("avg_humidity"),

        sum("rainfall").alias("total_rainfall"),

        avg("wind_speed").alias("avg_wind_speed"),

        avg("cloud_cover").alias("avg_cloud_cover")

    )

display(gold_df)
```

```python
gold_df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("gold.weather_summary")

print("Gold layer loaded successfully!")
```

```python
%sql
SELECT *
FROM gold.weather_summary
ORDER BY weather_date;
```
