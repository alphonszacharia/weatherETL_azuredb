```python
# Bronze count

bronze_count = spark.table(
    "bronze.weather_raw"
).count()

# Silver count

silver_count = spark.table(
    "silver.weather_cleaned"
).count()

# Gold count

gold_count = spark.table(
    "gold.weather_summary"
).count()


print("========== ETL VALIDATION ==========")

print("Bronze Records:", bronze_count)

print("Silver Records:", silver_count)

print("Gold Records:", gold_count)


assert bronze_count > 0, "Bronze table is empty"

assert silver_count > 0, "Silver table is empty"

assert gold_count > 0, "Gold table is empty"

print("ETL PIPELINE COMPLETED SUCCESSFULLY!")
```
